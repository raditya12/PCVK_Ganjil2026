```python
black_img = np.zeros(shape=(530,530,3),dtype=np.int16)
plt.imshow(black_img)
```

```python
cv.circle(
    black_img,
    center=(400,400),
    radius=50,
    color=(0,255,0),
    thickness=-1
)

cv.line(
    black_img,
    pt1=(0,0),
    pt2=(530,530),
    color=(255,255,255),
    thickness=5
)

plt.imshow(black_img)
```

```python
black_img2 = np.zeros(shape=(530,530,3),dtype=np.int16)
plt.imshow(black_img)
```

```python
vertices = np.array(
    [[150,350],
     [250,250],
     [450,350],
     [250,450]],
    dtype=np.int32
)

vertices
```

```python
pts = vertices.reshape((-1,1,2))
pts
```

```python
cv.polylines(
    black_img2,
    [pts],
    isClosed=True,
    color=(0,255,0),
    thickness=5
)

plt.imshow(black_img2)
```
