```python
import cv2 as cv
import matplotlib.pyplot as plt

# Membaca gambar
img = cv.imread('/content/drive/MyDrive/PCVK/KTM.jpg')

# Konversi BGR ke RGB agar benar saat ditampilkan dengan matplotlib
img_rgb = cv.cvtColor(img, cv.COLOR_BGR2RGB)

# Membuat tiga jenis pencerminan
flip_vertikal = cv.flip(img_rgb, 0)
flip_horizontal = cv.flip(img_rgb, 1)
flip_keduanya = cv.flip(img_rgb, -1)

# Membuat figure dengan 3 subplot
plt.figure(figsize=(15, 5))

# Vertikal
plt.subplot(1, 3, 1)
plt.imshow(flip_vertikal)
plt.title('Pencerminan Vertikal')
plt.axis('off')

# Horizontal
plt.subplot(1, 3, 2)
plt.imshow(flip_horizontal)
plt.title('Pencerminan Horizontal')
plt.axis('off')

# Vertikal + Horizontal
plt.subplot(1, 3, 3)
plt.imshow(flip_keduanya)
plt.title('Pencerminan Keduanya')
plt.axis('off')

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def tambah_label(foto):
    h, w = foto.shape[:2]


    hasil = np.zeros((h, w, 3), dtype=np.uint8)
    hasil[:, :] = foto[:, :]

    # Menghitung ukuran dan posisi secara adaptif
    tinggi_bar = max(1, h // 12)
    jarak_kiri = w // 50
    posisi_y = h - tinggi_bar // 3

    skala_font = min(tinggi_bar / 100, w / 700)
    ketebalan = max(1, round(skala_font * 2))

    # Menggambar bar hitam di bagian bawah
    cv2.rectangle(
        hasil,
        (0, h - tinggi_bar),
        (w - 1, h - 1),
        (0, 0, 0),
        -1
    )

    # Menambahkan label putih
    cv2.putText(
        hasil,
        "Dokumentasi PCVK - 2026",
        (jarak_kiri, posisi_y),
        cv2.FONT_HERSHEY_SIMPLEX,
        skala_font,
        (255, 255, 255),
        ketebalan,
        cv2.LINE_AA
    )

    return hasil



paths = [
    "/content/drive/MyDrive/PCVK/analisa1.png",
    "/content/drive/MyDrive/PCVK/analisa2.png"
]

# Matplotlib digunakan untuk menampilkan perbandingan
fig, axes = plt.subplots(2, 2)

for i, path in enumerate(paths):
    foto = cv2.imread(path)

    if foto is None:
        raise FileNotFoundError(f"Gambar tidak ditemukan: {path}")

    h, w = foto.shape[:2]

    # Fungsi yang sama diterapkan pada setiap gambar
    hasil = tambah_label(foto)

    # Menampilkan gambar masukan
    axes[i, 0].imshow(cv2.cvtColor(foto, cv2.COLOR_BGR2RGB))
    axes[i, 0].set_title(f"Masukan: {w} × {h} px")
    axes[i, 0].axis("off")

    # Menampilkan gambar keluaran
    axes[i, 1].imshow(cv2.cvtColor(hasil, cv2.COLOR_BGR2RGB))
    axes[i, 1].set_title(f"Keluaran: bar = {max(1, h // 12)} px")
    axes[i, 1].axis("off")

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def buat_thumbnail(foto, nama_toko="PCVK STORE"):
    h, w = foto.shape[:2]


    sisi = max(1, max(h, w) // 2)
    margin = sisi // 10
    ruang_foto = max(1, sisi - 2 * margin)

    # Skala yang sama untuk menjaga rasio lebar dan tinggi
    skala = min(ruang_foto / w, ruang_foto / h)
    lebar_baru = max(1, round(w * skala))
    tinggi_baru = max(1, round(h * skala))

    foto_resize = cv2.resize(
        foto,
        (lebar_baru, tinggi_baru),
        interpolation=cv2.INTER_AREA
    )


    kanvas = np.zeros((sisi, sisi, 3), dtype=np.uint8)
    kanvas[:, :] = (235, 225, 210)

    # Posisi foto di tengah kanvas
    x = (sisi - lebar_baru) // 2
    y = (sisi - tinggi_baru) // 2


    kanvas[y:y + tinggi_baru, x:x + lebar_baru] = foto_resize

    # Logo nama toko: posisi, ukuran, dan ketebalan adaptif
    posisi_logo = (sisi // 30, sisi - sisi // 30)
    skala_font = sisi / 1000
    ketebalan = max(1, round(sisi / 500))

    cv2.putText(
        kanvas,
        nama_toko,
        posisi_logo,
        cv2.FONT_HERSHEY_SIMPLEX,
        skala_font,
        (90, 60, 30),
        ketebalan,
        cv2.LINE_AA
    )

    return kanvas



paths = [
    "/content/drive/MyDrive/PCVK/ecommerce.png",
    "/content/drive/MyDrive/PCVK/ecommerce1.png"
]

# Tampilkan gambar asli dan thumbnail berdampingan
fig, axes = plt.subplots(2, 2)

for i, path in enumerate(paths):
    foto = cv2.imread(path)

    if foto is None:
        raise FileNotFoundError(
            f"Gambar tidak ditemukan: {path}. "
            "Pastikan Google Drive sudah terhubung."
        )

    thumbnail = buat_thumbnail(foto)

    h, w = foto.shape[:2]
    sisi = thumbnail.shape[0]

    axes[i, 0].imshow(cv2.cvtColor(foto, cv2.COLOR_BGR2RGB))
    axes[i, 0].set_title(f"Masukan: {w} × {h} px")
    axes[i, 0].axis("off")

    axes[i, 1].imshow(cv2.cvtColor(thumbnail, cv2.COLOR_BGR2RGB))
    axes[i, 1].set_title(f"Thumbnail: {sisi} × {sisi} px")
    axes[i, 1].axis("off")

plt.tight_layout()
plt.show()
```
